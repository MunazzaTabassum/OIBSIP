## 2. Missing Data Handling

The dataset contains missing values in the `Age` and `Salary` columns.

### Age
Age has 211 missing values. Since age is a numerical variable and can contain extreme values, median imputation is used. The median is less affected by potential outliers than the mean and provides a robust estimate of the typical employee age.

### Salary
Salary has 24 missing values. Since salary is a numerical variable that may be influenced by high earners, median imputation is also used. This avoids allowing extreme salary values to disproportionately affect the imputed values.

No rows are deleted because the missing values are limited to specific columns and can be reasonably imputed while preserving the available employee records.


```python
# Handle missing values using median imputation

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

print("Missing values after imputation:")
print(df[["Age", "Salary"]].isnull().sum())
```

### Missing Data Handling Decisions

The missing values were handled using column-specific strategies based on the nature of each variable.

- **Age:** Missing values were replaced using the median because Age is a numerical variable and the median is less affected by extreme values than the mean.
- **Salary:** Missing values were replaced using the median because Salary may contain unusually high or low values, making the median a more robust measure of central tendency.
- **Other columns:** Columns without missing values did not require imputation.

The selected strategies preserve the maximum number of employee records while reducing the effect of extreme values on the imputed results.

## 5. Outlier Detection

Outliers were identified in numerical variables using the Interquartile Range (IQR) method.

The IQR method identifies observations that fall below Q1 - 1.5 × IQR or above Q3 + 1.5 × IQR.

Age and Salary were examined separately because extreme values in these variables may affect downstream analysis and summary statistics.

```python
# Detect outliers using the IQR method

numeric_columns = ["Age", "Salary"]

outlier_summary = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

### Outlier Assessment

The IQR method was used to identify potential outliers in Age and Salary.

No outliers were detected in either variable. All employee records fall within the calculated IQR boundaries.

Therefore, no records were removed or modified based on outlier analysis. Retaining all observations preserves the complete employee population for downstream analysis.

## 7. Before vs. After Cleaning Summary

This section compares the original messy dataset with the cleaned dataset to evaluate the impact of the cleaning process.

The comparison includes:

- Row count
- Total null values
- Duplicate rows
- Data type accuracy

The purpose is to demonstrate that the cleaning process improved data quality while preserving valid employee records.

```python
# Load the original messy dataset for before-vs-after comparison

df_before = pd.read_csv("../dataset/Messy_Employee_dataset.csv")

# Before cleaning metrics
before_rows = len(df_before)
before_missing = df_before.isnull().sum().sum()
before_duplicates = df_before.duplicated().sum()


# Check expected data types using semantic type checks
def check_data_types(dataframe):

    checks = {
        "Employee_ID": is_text_column(dataframe["Employee_ID"]),
        "First_Name": is_text_column(dataframe["First_Name"]),
        "Last_Name": is_text_column(dataframe["Last_Name"]),
        "Age": is_float_dtype(dataframe["Age"]),
        "Department_Region": is_text_column(dataframe["Department_Region"]),
        "Status": is_text_column(dataframe["Status"]),
        "Join_Date": is_datetime64_any_dtype(dataframe["Join_Date"]),
        "Salary": is_float_dtype(dataframe["Salary"]),
        "Email": is_text_column(dataframe["Email"]),
        "Phone": is_text_column(dataframe["Phone"]),
        "Performance_Score": is_text_column(dataframe["Performance_Score"]),
        "Remote_Work": is_bool_dtype(dataframe["Remote_Work"])
    }

    return sum(checks.values())


# Count dtype matches
before_dtype_accuracy = check_data_types(df_before)
after_dtype_accuracy = check_data_types(df)


# After cleaning metrics
after_rows = len(df)
after_missing = df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()


# Create comparison table
comparison = pd.DataFrame({
    "Metric": [
        "Row Count",
        "Total Missing Values",
        "Duplicate Rows",
        "Correct Data Types"
    ],
    "Before Cleaning": [
        before_rows,
        before_missing,
        before_duplicates,
        f"{before_dtype_accuracy}/{len(df.columns)}"
    ],
    "After Cleaning": [
        after_rows,
        after_missing,
        after_duplicates,
        f"{after_dtype_accuracy}/{len(df.columns)}"
    ]
})

comparison
```
